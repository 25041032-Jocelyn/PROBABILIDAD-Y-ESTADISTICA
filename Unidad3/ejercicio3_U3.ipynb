```python
import matplotlib.pyplot as plt
import math
import numpy as np
```

```python
x = [0, 1, 2, 3, 4, 5]
p = [0.05, 0.10, 0.25, 0.30, 0.20, 0.10]
print ("Suma de probabilidades: ", sum(p))
```

```python
plt.stem(x, p)
plt.title("Distribución de Probabilidad")
plt.xlabel("Unidades vendidas")
plt.ylabel("Probabilidad")
plt.grid(True)
plt.show()
```

```python
valor_esperado = 0
for i in range (len(x)):
  xi = x[i]
  pi = p[i]
  producto = xi * pi
  valor_esperado += producto
  print (f"x = {xi}, p = {pi}, x * p = {producto:.2f}")
print(f"\nValor esperado de E(X): {valor_esperado:.2f}")
```

```python
esperanza = 0
for i in range (len(x)):
  esperanza += x[i] * p[i]
  esperanza_cuadrado = 0
  for i in range(len(x)):
    esperanza_cuadrado += (x[i] ** 2 * p[i])
varianza = esperanza_cuadrado - esperanza ** 2
desviacion = math.sqrt(varianza)

print(f"E(X) = {esperanza:.2f}")
print(f"E(X**2) = {esperanza_cuadrado:.2f}")
print(f"Varianza = {varianza:.2f}")
print(f"Desviación estándar = {desviacion:.2f}")
```

```python
F = np.cumsum(p)
for xi, fi in zip(x, F):
  print(f"F({xi}) = {fi:.2f}")

  plt.step(x, F, where = 'post')
  plt.title("Función de Distribución Acumulada (FDA)")
  plt.xlabel("Unidades vendidas")
  plt.ylabel("Probabilidad acumulada")
  plt.grid(True)
  plt.ylim(0, 1.05)
  plt.show()
```
