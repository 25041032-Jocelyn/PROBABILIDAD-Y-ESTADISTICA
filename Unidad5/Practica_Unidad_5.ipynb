```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from scipy.stats import pearsonr
```

```python
df1 = pd.read_csv("grupo1.csv")
df2 = pd.read_csv("grupo2.csv")
df3 = pd.read_csv("grupo3.csv")

print("Grupo 1")
print(df1.head())
print(df1.describe())

print("Grupo 2")
print(df2.head())
print(df2.describe())

print("Grupo 3")
print(df3.head())
print(df3.describe())
```

```python
#Grupo 1
horas_g1 = np.array(df1["HorasEstudio"]).reshape(-1, 1)
calificaciones_g1 = np.array(df1["Calificacion"])

modelo_g1 = LinearRegression()
modelo_g1.fit(horas_g1, calificaciones_g1)

print("Grupo 1")
print("Coeficiente de la pendiente: ", modelo_g1.coef_[0])
print("Intercepción de la recta: ", modelo_g1.intercept_)

coef_g1, p_g1 = pearsonr(df1["HorasEstudio"], df1["Calificacion"])
print("Coeficiente de correlación de Pearson: ", coef_g1)
print("Valor p: ", p_g1)
```

```python
plt.figure(figsize=(6, 4))
plt.hist(df1["Calificacion"], bins=10, color='steelblue', edgecolor='white')
plt.title("Histograma de Calificaciones Grupo 1")
plt.xlabel("Calificación Final")
plt.ylabel("Frecuencia")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.set(style='whitegrid')
sns.scatterplot(x=df1["HorasEstudio"], y=df1["Calificacion"])
plt.title("Diagrama de DIspersión Grupo 1")
plt.xlabel("Horas de Estudio")
plt.ylabel("Calificación Final")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.set(style='whitegrid')
sns.regplot(x=df1["HorasEstudio"], y=df1["Calificacion"], scatter=True, color='steelblue')
plt.title("Dispersión con Línea de Regresión Grupo 1")
plt.xlabel("Horas de Estudio")
plt.ylabel("Calificación Final")
plt.grid(True)
plt.show()
```

```python
#Grupo 2
horas_g2 = np.array(df2["HorasEstudio"]).reshape(-1, 1)
calificacion_g2 = np.array(df2["Calificacion"])

modelo_g2 = LinearRegression()
modelo_g2.fit(horas_g2, calificacion_g2)

print("\nGrupo 2")
print("Coeficiente de la pendiente: ", modelo_g2.coef_[0])
print("Interceoción de la recta: ", modelo_g2.intercept_)

coef_g2, p_g2 = pearsonr(df2["HorasEstudio"], df2["Calificacion"])
print("Coeficiente de correlación de Pearson: ", coef_g2)
print("Valor p: ", p_g2)
```

###¿Qué tan confiable es el modelo de regresión para predecir las calificaciones finales de los estudiantes basado en el tiempo de estudio?
Grupo 1 — R² = 0.0033 | RMSE = 7.74 puntos
Muy poco confiable. El modelo solo explica el 0.33% de la variación en las calificaciones. En la práctica, conocer las horas de estudio de un alumno de este grupo no te dice casi nada sobre su calificación. El error promedio de predicción es de ±7.74 puntos, lo cual es considerable.

Grupo 2 — R² = 0.7430 | RMSE = 10.37 puntos
Moderadamente confiable. El modelo explica el 74.3% de la variación en las calificaciones, lo que es bastante bueno. Sin embargo, el error promedio de ±10.37 puntos indica que las predicciones individuales pueden desviarse notablemente. Útil para tendencias generales, pero con reservas para predicciones individuales.

Grupo 3 — R² = 0.8878 | RMSE = 7.85 puntos
El más confiable de los tres. El modelo explica el 88.78% de la variación en las calificaciones. Es el grupo donde las horas de estudio predicen mejor el desempeño, con un error promedio de ±7.85 puntos, el más bajo junto con el Grupo 1 (pero aquí sí hay una relación real).

##Conclusiones
El análisis mostró resultados distintos entre los tres grupos. El Grupo 1 no presentó relación entre horas de estudio y calificaciones (r = -0.057, R² = 0.003, p = 0.69), por lo que el modelo no es útil para este grupo y el rendimiento probablemente depende de otros factores.
Los Grupos 2 y 3 sí mostraron una relación positiva fuerte y estadísticamente significativa (r = 0.862 y r = 0.942 respectivamente), con modelos que explican el 74.3% y 88.78% de la variación en las calificaciones. En ambos casos, estudiar más horas se asocia claramente con mejores calificaciones.
En conclusión, el tiempo de estudio es un buen predictor del rendimiento académico en los Grupos 2 y 3, siendo el Grupo 3 el más confiable, mientras que en el Grupo 1 se necesitarían variables adicionales para explicar el desempeño.
