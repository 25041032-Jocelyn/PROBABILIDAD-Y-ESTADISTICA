#**Distribución Binomial Y Poisson: Problema 1**

El problema nos indica que es la distribución de **Poisson**, también ya que contamos con eventos directos en un intervalo fijo de tiempo, hay tasa promedio constante, los eventos son independientes entre si y no hay un número fijo de intentos como en el binomial

```python
from scipy.stats import poisson
```

```python
x = 5

prob_a = poisson.pmf(2, x)
print(f"Probabilidad de que el número de personas ingresadas a una unidad de cuidados intensivos en un día sea dos = {prob_a:.6f} = {prob_a*100:.4f}%")

prob_leq_10 = poisson.cdf(10, x)
prob_b = 1 - prob_leq_10
print(f"\n  P(X > 10) = {prob_b:.6f} = {prob_b*100:.4f}%")
print(f"  P(X =< 10) = {prob_leq_10:.6f} = {prob_leq_10*100:.4f}%")

if prob_b < 0.05:
  print ("No es probable que x exceda 10 (probabilidad < 5%)")
else:
  print("Si es probable que x exceda 10 (probabilidad >= 5%)")
```
