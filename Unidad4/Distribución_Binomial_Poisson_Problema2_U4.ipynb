#**Distribución Binomial Y Poisson: Problema 2**

El problema nos da indicios de ser Binomial debido a que usa n ensayos fijos, cada ensayo es éxito o fracaso, los ensayos son independientes y p es constante, sin embargo, n es muy grande y p muy pequeño, lo cual es complicado de resolver sin computadora. **Poisson** nos ofrece una forma más sencilla de obtener el resultado.

```python
from scipy.stats import poisson
```

```python
n = 1000 #lote de motores
p = 0.001 #prob de motor defectuoso
lam = n * p

prob_a = poisson.pmf(0, lam)
print(f"Probabilidad de que ningún motor sea defectuoso = {prob_a:.6f} = {prob_a*100:.4f}%")

prob_b = poisson.pmf(3, lam)
print(f"Probabilidad de que 3 motores sean defectuosos = {prob_b:.6f} = {prob_b*100:.4f}%")

prob_c = poisson.pmf(4, lam)
print(f"Probabilidad de que 4 sean motores sean defectuosos = {prob_c:.6f} = {prob_c*100:.4f}%")
```
