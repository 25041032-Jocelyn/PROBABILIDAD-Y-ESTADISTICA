```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
media = 70
desv_std = 4
total_estudiantes = 500

peso_limite_menos = 68
peso_limite_mas = 72

prob_hasta_68kg = norm.cdf(peso_limite_menos, loc=media, scale=desv_std)
prob_hasta_72kg = norm.cdf(peso_limite_mas, loc=media, scale=desv_std)

prob_entre_68_72kg = prob_hasta_72kg - prob_hasta_68kg

estudiantes_entre_68_72kg = prob_entre_68_72kg * total_estudiantes

print("Probabilidad de que un estudiante pese más de 68 kg y menos de 72 kg: ", prob_entre_68_72kg)
print("Número de estudiantes que pesarán entre 68 kg y 72 kg: ", estudiantes_entre_68_72kg)
```

```python
x = np.linspace(media - 4 * desv_std, media + 4 * desv_std, 1000)
y = norm.pdf(x, media, desv_std)

plt.plot(x, y, 'b', linewidth=2)
plt.fill_between(x, y, where=(x >= peso_limite_menos) & (x <= peso_limite_mas), color='skyblue', alpha=0.5)
plt.title(r'Probabilidad de que un estudiante pese más de 62 kg y menos de 72 kg. ' + '\n' + r'$\mu=70$, $\sigma=4$, X=500')
plt.xlabel('Peso (kg)')
plt.ylabel('Densidad de probabilidad')
plt.grid(True)
plt.show()
```
